Esta libreta oficial ejecuta el **nuevo pipeline cinemático V3 de inicio a fin en 1 solo clic** en la infraestructura de alta velocidad de Google Colab. No consume CPU, memoria RAM ni tarjeta gráfica de tu computadora local.

### 🌟 Novedades Visuales & Mejoras de Producción V3:
- 🎯 **Flechas Vectoriales Animadas (SVG):** Curvas Bézier dinámicas que se auto-dibujan en pantalla (`stroke-dashoffset`) con punta elástica esmeralda neón (`#10b981`), apuntando con precisión milimétrica a los botones reales.
- 💎 **La Demostración del 'Después' (Value Payoff):** La página no se ve vacía. Al crear la cuenta, estalla el confeti nativo (`canvas-confetti`), se activa la tarjeta institucional de $100K y se revela el Dashboard poblado con datos reales: curva de equidad ascendente, Win Rate del 68.4%, Profit Factor de 2.35 y calendario verde.
- 🛡️ **Cero Fotogramas Blancos:** Protocolo CDP para arrancar Chromium directamente en fondo oscuro `#0b0f19` antes de procesar el primer frame.
- 🎞️ **Transiciones Cinemáticas Suaves:** Disoluciones cruzadas en FFmpeg (`xfade=transition=fade:duration=0.5`) entre bloques sin cortes bruscos.
- 🎙️ **Locución Modular Jorge Neural:** 4 bloques de audio con ritmo institucional y dicción técnica impecable.
- 💾 **Entrega Directa en Google Drive:** Video final 1080p MP4 guardado en `/MyDrive/TradingJournalPro_Tutorials_YouTube/`.

```python
#@title 🚀 EJECUCIÓN MAESTRA 1-CLICK (Tutorial #1 AsTrade V3) { run: "auto" }
#@markdown Pulsa el botón de Play (▶️). Todo se ejecutará de forma 100% autónoma en los servidores de Google.

modo_renderizado = "Tutorial 1: Onboarding y Cuenta Challenge $100K (V3 Motion Graphics)" #@param ["Tutorial 1: Onboarding y Cuenta Challenge $100K (V3 Motion Graphics)"]
forzar_re_renderizado = False #@param {type:"boolean"}

import os
import sys
import json
import time
import shutil
import zipfile
import subprocess
from google.colab import drive

t0_global = time.time()
print("=" * 70)
print("🚀 INICIANDO ESTUDIO MAESTRO DE PRODUCCIÓN V3 — ASTRADE")
print("🎬 Tutorial #1: Onboarding y Creación de Cuenta Challenge $100K")
print("=" * 70)

# 1. Conectar Google Drive
print("\n[1/6] 📁 Conectando con Google Drive...")
drive.mount('/content/drive')
DRIVE_MYDRIVE = '/content/drive/MyDrive'
DRIVE_OUT = os.path.join(DRIVE_MYDRIVE, 'TradingJournalPro_Tutorials_YouTube')
os.makedirs(DRIVE_OUT, exist_ok=True)
print(f"  -> Carpeta de entrega en Drive: {DRIVE_OUT}")

# 2. Localizar y preparar el proyecto
print("\n[2/6] 📦 Preparando archivos del proyecto...")
TARGET_DIR = '/content/trading-saas'
DRIVE_BUNDLE_V3 = os.path.join(DRIVE_MYDRIVE, 'astrade_video01_bundle.zip')
DRIVE_BUNDLE_LEGACY = os.path.join(DRIVE_MYDRIVE, 'trading_saas_bundle.zip')
DRIVE_FOLDER = os.path.join(DRIVE_MYDRIVE, 'trading-saas')

if os.path.exists(DRIVE_BUNDLE_V3):
    print(f"  -> Encontrado paquete V3 en Drive: {DRIVE_BUNDLE_V3}")
    if os.path.exists(TARGET_DIR):
        shutil.rmtree(TARGET_DIR, ignore_errors=True)
    os.makedirs(TARGET_DIR, exist_ok=True)
    with zipfile.ZipFile(DRIVE_BUNDLE_V3, 'r') as zf:
        zf.extractall(TARGET_DIR)
elif os.path.exists(DRIVE_BUNDLE_LEGACY):
    print(f"  -> Encontrado paquete en Drive: {DRIVE_BUNDLE_LEGACY}")
    if os.path.exists(TARGET_DIR):
        shutil.rmtree(TARGET_DIR, ignore_errors=True)
    os.makedirs(TARGET_DIR, exist_ok=True)
    with zipfile.ZipFile(DRIVE_BUNDLE_LEGACY, 'r') as zf:
        zf.extractall(TARGET_DIR)
elif os.path.exists(DRIVE_FOLDER):
    print(f"  -> Encontrada carpeta de proyecto en Drive: {DRIVE_FOLDER}")
    if os.path.exists(TARGET_DIR):
        shutil.rmtree(TARGET_DIR, ignore_errors=True)
    shutil.copytree(DRIVE_FOLDER, TARGET_DIR, dirs_exist_ok=True)
else:
    print("  -> Descargando / Clonando repositorio desde GitHub...")
    if not os.path.exists(TARGET_DIR):
        subprocess.run(['git', 'clone', 'https://github.com/juankudo/trading-saas.git', TARGET_DIR], check=True)
    else:
        subprocess.run(['git', '-C', TARGET_DIR, 'pull'])

%cd {TARGET_DIR}
print(f"  -> Directorio de trabajo activo: {os.getcwd()}")

# 3. Instalación de Dependencias del Sistema & Python
print("\n[3/6] ⚙️ Verificando e instalando dependencias multimedia...")
subprocess.run(['apt-get', 'update', '-qq'], stdout=subprocess.DEVNULL)
subprocess.run(['apt-get', 'install', '-y', '-qq', 'ffmpeg', 'xvfb', 'libasound2', 'libgbm1'], stdout=subprocess.DEVNULL)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'playwright', 'edge-tts', 'imageio-ffmpeg', 'pillow', 'mutagen', 'tqdm'], check=True)
subprocess.run(['playwright', 'install', 'chromium'], stdout=subprocess.DEVNULL)
subprocess.run(['playwright', 'install-deps', 'chromium'], stdout=subprocess.DEVNULL)

import imageio_ffmpeg
FFMPEG_EXE = imageio_ffmpeg.get_ffmpeg_exe()
print(f"  -> FFmpeg binario activo: {FFMPEG_EXE}")

# 4. Verificación y Auditoría de Audios Modulares V3
print("\n[4/6] 🎙️ Auditando audios modulares V3 (JorgeNeural)...")
MANIFEST_V3 = os.path.join(TARGET_DIR, 'scratch', 'video_01_v3', 'audio', 'manifest_v3.json')
if not os.path.exists(MANIFEST_V3):
    print("  -> Generando audios de locución V3 con Edge-TTS...")
    subprocess.run([sys.executable, os.path.join(TARGET_DIR, 'scripts', 'audio', 'generate_video_01_v3_audio.py')], check=True)
else:
    print(f"  -> Audios V3 verificados y listos en {MANIFEST_V3}")

# 5. Checkpoint en Google Drive
print("\n[5/6] 🛡️ Verificando Checkpoint en Google Drive...")
FINAL_DRIVE_MP4 = os.path.join(DRIVE_OUT, 'AsTrade_Tutorial_01_Onboarding_1080p.mp4')
if not forzar_re_renderizado and os.path.exists(FINAL_DRIVE_MP4) and os.path.getsize(FINAL_DRIVE_MP4) > 1000000:
    print(f"  ⏩ CHECKPOINT ACTIVO: El video ya existe en Drive ({os.path.getsize(FINAL_DRIVE_MP4)/(1024*1024):.2f} MB).")
    print("     (Activa 'forzar_re_renderizado' si deseas sobreescribirlo).")
else:
    # 6. Ejecución del Renderizado Cinemático V3 (Aislado en Subproceso)
    print("\n[6/6] 🎬 Ejecutando renderizado cinemático con Motion Graphics...")
    RENDER_SCRIPT = os.path.join(TARGET_DIR, 'scripts', 'render', 'render_video_01_astrade.py')
    
    # Subproceso independiente: 100% inmune al bucle asyncio de Jupyter/Colab
    proc = subprocess.Popen([sys.executable, RENDER_SCRIPT], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in proc.stdout:
        print(line, end='', flush=True)
    proc.wait()

    LOCAL_MP4 = os.path.join(TARGET_DIR, 'media', 'youtube_tutorials', 'AsTrade_Tutorial_01_Onboarding_1080p.mp4')
    if os.path.exists(LOCAL_MP4) and os.path.getsize(LOCAL_MP4) > 1000000:
        shutil.copy2(LOCAL_MP4, FINAL_DRIVE_MP4)
        print(f"\n💾 ¡GUARDADO EN TU GOOGLE DRIVE!")
        print(f"   📍 Ruta: {FINAL_DRIVE_MP4}")
        print(f"   📦 Tamaño: {os.path.getsize(FINAL_DRIVE_MP4)/(1024*1024):.2f} MB")
    else:
        print(f"\n⚠️ Nota: No se encontró el video generado en {LOCAL_MP4}. Revisa los logs de arriba.")

elapsed_tot = time.time() - t0_global
print("\n" + "=" * 70)
print("🎉 ¡PROCESO DE PRODUCCIÓN V3 COMPLETADO!")
print(f"⏱️ Tiempo total: {elapsed_tot/60:.2f} minutos")
print(f"📁 Carpeta de destino en Google Drive: {DRIVE_OUT}")
print("=" * 70)
```
